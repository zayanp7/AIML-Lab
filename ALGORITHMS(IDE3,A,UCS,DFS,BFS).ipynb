### A* (A-Star) Search Algorithm
A* is a popular pathfinding algorithm that finds the shortest path between a starting node and a target node in a graph. It uses a heuristic function to estimate the cost from the current node to the goal, making it more efficient than other uninformed search algorithms like Dijkstra's. A* evaluates nodes using the function `f(n) = g(n) + h(n)`, where:
- `g(n)` is the cost from the start node to node `n`.
- `h(n)` is the heuristic estimate of the cost from node `n` to the goal node.

```python
import heapq

def a_star_search(graph, start, goal, heuristic):
    """
    Implements the A* search algorithm.

    Args:
        graph (dict): A dictionary representing the graph where keys are nodes
                      and values are dictionaries of neighbors and their costs.
                      Example: {'A': {'B': 1, 'C': 4}, 'B': {'D': 2}, ...}
        start: The starting node.
        goal: The goal node.
        heuristic (dict): A dictionary mapping nodes to their heuristic estimates
                          to the goal.

    Returns:
        tuple: A tuple containing the path (list) and its total cost (int/float)
               if a path is found, otherwise (None, float('inf')).
    """
    open_set = []
    heapq.heappush(open_set, (0 + heuristic[start], 0, start, [start]))

    g_score = {node: float('inf') for node in graph}
    g_score[start] = 0

    while open_set:
        f_score, current_cost, current_node, path = heapq.heappop(open_set)

        if current_node == goal:
            return path, current_cost

        for neighbor, weight in graph.get(current_node, {}).items():
            tentative_g_score = current_cost + weight

            if tentative_g_score < g_score[neighbor]:
                g_score[neighbor] = tentative_g_score
                f_score = tentative_g_score + heuristic[neighbor]
                heapq.heappush(open_set, (f_score, tentative_g_score, neighbor, path + [neighbor]))

    return None, float('inf')
```

```python
# Example Usage for A* Algorithm
print("\n--- A* Algorithm Example ---")
graph_a_star = {
    'A': {'B': 1, 'C': 4},
    'B': {'D': 2, 'E': 5},
    'C': {'F': 1},
    'D': {'G': 3},
    'E': {'G': 1},
    'F': {'G': 1},
    'G': {}
}

heuristic_a_star = {
    'A': 7,
    'B': 6,
    'C': 2,
    'D': 2,
    'E': 1,
    'F': 1,
    'G': 0
}

start_node_a_star = 'A'
goal_node_a_star = 'G'

path_a_star, cost_a_star = a_star_search(graph_a_star, start_node_a_star, goal_node_a_star, heuristic_a_star)

if path_a_star:
    print(f"A* Path from {start_node_a_star} to {goal_node_a_star}: {path_a_star}")
    print(f"A* Total Cost: {cost_a_star}")
else:
    print(f"A* Path from {start_node_a_star} to {goal_node_a_star} not found.")
```

### Uniform Cost Search (UCS)
Uniform Cost Search (UCS) is a tree search algorithm used for traversing or searching a weighted graph to find the shortest path from a starting node to a target node. It explores the node with the lowest cumulative cost from the start node first. UCS is essentially a special case of Dijkstra's algorithm or A* algorithm where the heuristic function `h(n)` is always zero. It guarantees finding the path with the lowest cost.

```python
import heapq

def uniform_cost_search(graph, start, goal):
    """
    Implements the Uniform Cost Search (UCS) algorithm.

    Args:
        graph (dict): A dictionary representing the graph where keys are nodes
                      and values are dictionaries of neighbors and their costs.
                      Example: {'A': {'B': 1, 'C': 4}, 'B': {'D': 2}, ...}
        start: The starting node.
        goal: The goal node.

    Returns:
        tuple: A tuple containing the path (list) and its total cost (int/float)
               if a path is found, otherwise (None, float('inf')).
    """
    priority_queue = [(0, start, [start])]

    g_score = {node: float('inf') for node in graph}
    g_score[start] = 0

    while priority_queue:
        cost, current_node, path = heapq.heappop(priority_queue)

        if current_node == goal:
            return path, cost

        if cost > g_score[current_node]:
            continue

        for neighbor, weight in graph.get(current_node, {}).items():
            new_cost = cost + weight

            if new_cost < g_score[neighbor]:
                g_score[neighbor] = new_cost
                heapq.heappush(priority_queue, (new_cost, neighbor, path + [neighbor]))

    return None, float('inf')
```

```python
# Example Usage for Uniform Cost Search
print("\n--- Uniform Cost Search Example ---")
graph_ucs = {
    'A': {'B': 1, 'C': 5},
    'B': {'D': 2, 'E': 4},
    'C': {'F': 3},
    'D': {'G': 3},
    'E': {'G': 1},
    'F': {'G': 1},
    'G': {}
}

start_node_ucs = 'A'
goal_node_ucs = 'G'

path_ucs, cost_ucs = uniform_cost_search(graph_ucs, start_node_ucs, goal_node_ucs)

if path_ucs:
    print(f"UCS Path from {start_node_ucs} to {goal_node_ucs}: {path_ucs}")
    print(f"UCS Total Cost: {cost_ucs}")
else:
    print(f"UCS Path from {start_node_ucs} to {goal_node_ucs} not found.")
```

### Greedy Best-First Search
Greedy Best-First Search (GBFS) is an informed search algorithm that attempts to find the shortest path from a start node to a goal node. Unlike A*, GBFS expands nodes based solely on the heuristic function `h(n)`, which estimates the cost from the current node to the goal. It prioritizes nodes that appear to be closest to the goal, making it 'greedy' in its approach.

Key characteristics:
- **Heuristic-driven**: Only uses the heuristic function `h(n)` to decide which node to explore next.
- **Not optimal**: Does not guarantee finding the shortest path, as it might get trapped in local optima.
- **Faster**: Can be faster than A* or UCS in certain scenarios because it doesn't consider the cost from the start node (`g(n)`).
- **Complete**: Is complete if the state space is finite and there are no loops.

```python
import heapq

def greedy_best_first_search(graph, start, goal, heuristic):
    """
    Implements the Greedy Best-First Search algorithm.

    Args:
        graph (dict): A dictionary representing the graph where keys are nodes
                      and values are dictionaries of neighbors and their costs.
                      Example: {'A': {'B': 1, 'C': 4}, 'B': {'D': 2}, ...}
        start: The starting node.
        goal: The goal node.
        heuristic (dict): A dictionary mapping nodes to their heuristic estimates
                          to the goal.

    Returns:
        tuple: A tuple containing the path (list) and its total cost (int/float)
               if a path is found, otherwise (None, float('inf')).
               Note: GBFS finds a path, but not necessarily the optimal cost.
                     The cost returned is the sum of edge weights along the found path.
    """
    priority_queue = [(heuristic[start], 0, start, [start])]
    visited = set()

    while priority_queue:
        h_score, current_cost, current_node, path = heapq.heappop(priority_queue)

        if current_node == goal:
            return path, current_cost

        if current_node in visited:
            continue

        visited.add(current_node)

        for neighbor, weight in graph.get(current_node, {}).items():
            if neighbor not in visited:
                new_cost = current_cost + weight
                heapq.heappush(priority_queue, (heuristic[neighbor], new_cost, neighbor, path + [neighbor]))

    return None, float('inf')
```

```python
# Example Usage for Greedy Best-First Search
print("\n--- Greedy Best-First Search Example ---")
graph_gbfs = {
    'A': {'B': 1, 'C': 4},
    'B': {'D': 2, 'E': 5},
    'C': {'F': 1},
    'D': {'G': 3},
    'E': {'G': 1},
    'F': {'G': 1},
    'G': {}
}

heuristic_gbfs = {
    'A': 7,
    'B': 6,
    'C': 2,
    'D': 2,
    'E': 1,
    'F': 1,
    'G': 0
}

start_node_gbfs = 'A'
goal_node_gbfs = 'G'

path_gbfs, cost_gbfs = greedy_best_first_search(graph_gbfs, start_node_gbfs, goal_node_gbfs, heuristic_gbfs)

if path_gbfs:
    print(f"GBFS Path from {start_node_gbfs} to {goal_node_gbfs}: {path_gbfs}")
    print(f"GBFS Total Cost: {cost_gbfs}")
else:
    print(f"GBFS Path from {start_node_gbfs} to {goal_node_gbfs} not found.")
```

### Breadth-First Search (BFS)
Breadth-First Search (BFS) is an algorithm for traversing or searching tree or graph data structures. It starts at the tree root (or some arbitrary node of a graph, sometimes referred to as a 'search key') and explores all of the neighbor nodes at the present depth prior to moving on to the nodes at the next depth level.

Key characteristics:
- **Completeness**: BFS is complete; if there is a solution, BFS is guaranteed to find it.
- **Optimality**: If the graph is unweighted (or all edge weights are equal), BFS is optimal (finds the shortest path in terms of number of edges).
- **Time Complexity**: O(V + E), where V is the number of vertices and E is the number of edges.
- **Space Complexity**: O(V), as it needs to store all nodes in the current level.

```python
from collections import deque

def breadth_first_search(graph, start, goal):
    """
    Implements the Breadth-First Search (BFS) algorithm.

    Args:
        graph (dict): A dictionary representing the graph where keys are nodes
                      and values are lists of neighbors.
                      Example: {'A': ['B', 'C'], 'B': ['D', 'E'], ...}
        start: The starting node.
        goal: The goal node.

    Returns:
        tuple: A tuple containing the path (list) and its total cost (int) (number of edges)
               if a path is found, otherwise (None, float('inf')).
    """
    queue = deque([(start, [start], 0)])  # (current_node, path, cost)
    visited = {start}

    while queue:
        current_node, path, cost = queue.popleft()

        if current_node == goal:
            return path, cost

        for neighbor in graph.get(current_node, []):
            if neighbor not in visited:
                visited.add(neighbor)
                queue.append((neighbor, path + [neighbor], cost + 1))

    return None, float('inf')
```

```python
# Example Usage for Breadth-First Search
print("\n--- Breadth-First Search Example ---")
graph_bfs = {
    'A': ['B', 'C'],
    'B': ['D', 'E'],
    'C': ['F'],
    'D': ['G'],
    'E': ['G'],
    'F': ['G'],
    'G': []
}

start_node_bfs = 'A'
goal_node_bfs = 'G'

path_bfs, cost_bfs = breadth_first_search(graph_bfs, start_node_bfs, goal_node_bfs)

if path_bfs:
    print(f"BFS Path from {start_node_bfs} to {goal_node_bfs}: {path_bfs}")
    print(f"BFS Total Cost (edges): {cost_bfs}")
else:
    print(f"BFS Path from {start_node_bfs} to {goal_node_bfs} not found.")
```

### Depth-First Search (DFS)
Depth-First Search (DFS) is an algorithm for traversing or searching tree or graph data structures. The algorithm starts at the root node (selecting some arbitrary node as the root in the case of a graph) and explores as far as possible along each branch before backtracking.

Key characteristics:
- **Completeness**: DFS is complete if the graph is finite and does not contain cycles, or if cycles are handled explicitly (e.g., by keeping track of visited nodes).
- **Optimality**: DFS is not optimal; it may find a very long path to the goal, even if a shorter path exists.
- **Time Complexity**: O(V + E), where V is the number of vertices and E is the number of edges.
- **Space Complexity**: O(V) in the worst case (for storing the stack or recursion call stack), but can be much smaller for shallow trees.

```python
def depth_first_search(graph, start, goal):
    """
    Implements the Depth-First Search (DFS) algorithm.

    Args:
        graph (dict): A dictionary representing the graph where keys are nodes
                      and values are lists of neighbors.
                      Example: {'A': ['B', 'C'], 'B': ['D', 'E'], ...}
        start: The starting node.
        goal: The goal node.

    Returns:
        tuple: A tuple containing the path (list) and its total cost (int) (number of edges)
               if a path is found, otherwise (None, float('inf')).
    """
    stack = [(start, [start], 0)]  # (current_node, path, cost)
    visited = set()

    while stack:
        current_node, path, cost = stack.pop()

        if current_node == goal:
            return path, cost

        if current_node not in visited:
            visited.add(current_node)
            for neighbor in reversed(graph.get(current_node, [])): # Reverse to explore in a consistent order
                if neighbor not in visited:
                    stack.append((neighbor, path + [neighbor], cost + 1))

    return None, float('inf')
```

```python
# Example Usage for Depth-First Search
print("\n--- Depth-First Search Example ---")
graph_dfs = {
    'A': ['B', 'C'],
    'B': ['D', 'E'],
    'C': ['F'],
    'D': ['G'],
    'E': [],
    'F': ['G'],
    'G': []
}

start_node_dfs = 'A'
goal_node_dfs = 'G'

path_dfs, cost_dfs = depth_first_search(graph_dfs, start_node_dfs, goal_node_dfs)

if path_dfs:
    print(f"DFS Path from {start_node_dfs} to {goal_node_dfs}: {path_dfs}")
    print(f"DFS Total Cost (edges): {cost_dfs}")
else:
    print(f"DFS Path from {start_node_dfs} to {goal_node_dfs} not found.")
```

### ID3 (Iterative Dichotomiser 3) Algorithm

ID3 is a classic algorithm used to construct a decision tree from a dataset. It's a greedy algorithm that builds a tree top-down, selecting the best attribute at each node to split the data. The 'best' attribute is determined by its **Information Gain**, which measures how much entropy (disorder) is reduced by splitting the dataset on that attribute.

**Key Concepts:**

*   **Entropy**: A measure of impurity or disorder in a set of examples. For a set $S$ with $c$ classes, the entropy is calculated as:

    $$H(S) = -\sum_{i=1}^{c} p_i \log_2(p_i)$$

    where $p_i$ is the proportion of examples belonging to class $i$.

*   **Information Gain**: The reduction in entropy achieved by splitting the dataset $S$ on an attribute $A$. It's calculated as:

    $$Gain(S, A) = H(S) - \sum_{v \in Values(A)} \frac{|S_v|}{|S|} H(S_v)$$

    where $Values(A)$ is the set of all possible values for attribute $A$, $S_v$ is the subset of $S$ for which attribute $A$ has value $v$, and $|S_v| / |S|$ is the proportion of examples in $S_v$.

**Algorithm Steps:**

1.  **Start with the entire dataset $S$ as the root node.**
2.  **Calculate the entropy of the target attribute for $S$.**
3.  **For each attribute, calculate the information gain.**
4.  **Select the attribute with the highest information gain as the splitting attribute for the current node.**
5.  **Create child nodes for each unique value of the chosen attribute.**
6.  **Partition the dataset into subsets based on the chosen attribute's values.**
7.  **Recursively apply steps 2-6 to each child node's subset until:**
    *   All examples in a subset belong to the same class (a leaf node).
    *   There are no more attributes to split on (a leaf node with the majority class).
    *   There are no more examples in the subset.

```python
import pandas as pd
import math
import numpy as np

def entropy(target_col):
    """
    Calculates the entropy of a given target column.
    """
    elements, counts = np.unique(target_col, return_counts=True)
    entropy_val = -sum([(counts[i]/len(target_col)) * math.log2(counts[i]/len(target_col))
                        for i in range(len(elements))])
    return entropy_val

def information_gain(data, feature_name, target_name):
    """
    Calculates the information gain of a feature.
    """
    # Calculate entropy of the total dataset
    total_entropy = entropy(data[target_name])

    # Calculate entropy for each unique value of the feature
    vals, counts = np.unique(data[feature_name], return_counts=True)
    weighted_entropy = sum([(counts[i]/len(data)) * entropy(data.where(data[feature_name]==vals[i]).dropna()[target_name])
                            for i in range(len(vals))])

    # Calculate information gain
    gain = total_entropy - weighted_entropy
    return gain

def id3_algorithm(data, original_data, features, target_attribute_name, parent_node_class=None):
    """
    Implements the ID3 algorithm to build a decision tree.
    """
    # If all target values are the same, return that value
    if len(np.unique(data[target_attribute_name])) <= 1:
        return np.unique(data[target_attribute_name])[0]

    # If dataset is empty, return parent_node_class
    elif len(data) == 0:
        return parent_node_class

    # If no more features to split on, return the most common target value
    elif len(features) == 0:
        return original_data[target_attribute_name].value_counts().idxmax()

    else:
        # Determine the default class (most common in the current data)
        parent_node_class = data[target_attribute_name].value_counts().idxmax()

        # Select the feature with the highest information gain
        item_values = [information_gain(data, feature, target_attribute_name) for feature in features]
        best_feature_index = np.argmax(item_values)
        best_feature = features[best_feature_index]

        # Create the tree structure
        tree = {best_feature: {}}

        # Remove the chosen feature from the feature list
        features = [i for i in features if i != best_feature]

        # Build tree for each unique value of the best feature
        for value in np.unique(data[best_feature]):
            value = value
            sub_data = data.where(data[best_feature] == value).dropna()
            subtree = id3_algorithm(sub_data, original_data, features, target_attribute_name, parent_node_class)
            tree[best_feature][value] = subtree

        return tree
```

```python
# Example Usage for ID3 Algorithm
import pandas as pd
print("\n--- ID3 Algorithm Example ---")

# Sample dataset (Play Tennis example from ID3 literature)
data = {
    'Outlook': ['Sunny', 'Sunny', 'Overcast', 'Rain', 'Rain', 'Rain', 'Overcast', 'Sunny', 'Sunny', 'Rain', 'Sunny', 'Overcast', 'Overcast', 'Rain'],
    'Temperature': ['Hot', 'Hot', 'Hot', 'Mild', 'Cool', 'Cool', 'Cool', 'Mild', 'Cool', 'Mild', 'Mild', 'Mild', 'Hot', 'Mild'],
    'Humidity': ['High', 'High', 'High', 'High', 'Normal', 'Normal', 'Normal', 'High', 'Normal', 'Normal', 'Normal', 'High', 'Normal', 'High'],
    'Wind': ['Weak', 'Strong', 'Weak', 'Weak', 'Weak', 'Strong', 'Strong', 'Weak', 'Weak', 'Weak', 'Strong', 'Strong', 'Weak', 'Strong'],
    'PlayTennis': ['No', 'No', 'Yes', 'Yes', 'Yes', 'No', 'Yes', 'No', 'Yes', 'Yes', 'Yes', 'Yes', 'Yes', 'No']
}
df = pd.DataFrame(data)

# Define features and target attribute
features = df.columns.drop('PlayTennis').tolist()
target = 'PlayTennis'

# Build the decision tree using ID3
decision_tree = id3_algorithm(df, df, features, target)

# Print the resulting decision tree
import json
print("\nDecision Tree:")
print(json.dumps(decision_tree, indent=4))

# You can also add a function to classify a new instance using the tree
def classify(instance, tree):
    attribute = list(tree.keys())[0]
    value = instance[attribute]
    if value in tree[attribute]:
        next_level = tree[attribute][value]
        if isinstance(next_level, dict):
            return classify(instance, next_level)
        else:
            return next_level
    else:
        return "Unknown"

# Example of classifying a new instance
new_instance = {'Outlook': 'Sunny', 'Temperature': 'Cool', 'Humidity': 'High', 'Wind': 'Strong'}
prediction = classify(new_instance, decision_tree)
print(f"\nClassifying new instance {new_instance}: {prediction}")

new_instance_2 = {'Outlook': 'Overcast', 'Temperature': 'Hot', 'Humidity': 'Normal', 'Wind': 'Weak'}
prediction_2 = classify(new_instance_2, decision_tree)
print(f"Classifying new instance {new_instance_2}: {prediction_2}")
```
